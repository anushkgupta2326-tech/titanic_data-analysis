```python
# Create a student record management

# Create a list with student id, student name, dept, list of subjects, list of marks

students = [
    [101, "Anish", "CSE", ["Python", "Java", "DBMS"], [85, 78, 90]]
]

# Display all the items in the list
print(students)

# Arrange the subject names in ascending order
students[0][3].sort()
print("Subjects:", students[0][3])

# Find the maximum mark and display
print("Maximum mark:", max(students[0][4]))

# Display minimum mark
print("Minimum mark:", min(students[0][4]))

# Display average mark
print("Average mark:", sum(students[0][4]) / len(students[0][4]))
```

# A tuple is a collection of items
# It is immutable
# Tuple is given in ()

```python
# Create a tuple

student = ('Chirag', 21, 'CSE', 8.7)

print(student)
```

```python
# Display the items using index

subjects = ('Python', 'DSA', 'DAA', 'OS', 'SQL', 'Statistics')

print(subjects[0])
print(subjects[5])
```

```python
subjects = ('Python', 'DSA', 'DAA', 'OS', 'SQL', 'Statistics')
subjects[2]="Algebra"
```

```python
# Tuple slicing

subjects = ('Python', 'DSA', 'DAA', 'OS', 'SQL', 'Statistics')

print(subjects[1:4])
print ("No of subjects:",len(subjects))
```

```python
# Nested tuple

student = (101, 'Rahul', (85, 88, 76))
print(student)
print(student[2])
print(student[2][1])
```

```python
# Tuple slicing, length and finding an item
subjects = ('Python', 'Statistics', 'SQL', 'Machine Learning', 'Deep Learning')

print(subjects[1:4])
print("Number of Subjects:", len(subjects))
print("Deep Learning" in subjects)
print("Relational Algebra" in subjects)
```
