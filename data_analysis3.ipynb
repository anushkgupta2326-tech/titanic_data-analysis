Tuple

```python
# create a tuple
student=(181,"sanjeev","CSE",8.9)
print(student)
```

```python
#tuple by using the index
print("student information in first location: [0]",student[0])
print(student[1])
```

```python
#negative index
print("Example for negative index:",student[-2])
```

```python
#to find the length of the tuple
subjects=("dsa","dsv","os","python","machine learning")
print(len(subjects))
```

```python
#slicing
subjects=("dsa","dsv","os","python","machine learning")
print(subjects[1:3])
print(subjects[:5])
print(subjects[4:])
```

```python
#Tuple is immutable
students_info=("chirag",21,"CSE",8.7)
students_info[3]=9.1
print (students_info)
```

```python
#nested tuple
student=(101,"Rahul",(81,78,66))
print(student)
print(student[2])
print(student[2][1])
```

```python
#find the element present in tuple or not
subjects=("python","DSV","SQL","Machine learning","statistics")
print("python" in subjects)
print("java" in subjects)
```

```python
# SET
## unordered collection of items
## mutable
## allow the different data types
## it doesn't support for index
## it gives output with unique (to duplication)
```

```python
#Symbol for set { }
#create a set with simmilar items
marks={75,68,89,55,78,89,75,66,55,99}
print(marks)
```

```python
#create a empty set
a={}
a=set()
#check its types
print(type(a))
```

```python
# set doesn't support the index
students= {"Amit","Riya","priya"}
print (students[1])
```

```python
# add an item in the set
students={"Amit","kiran","sneha"}
students.add("priya")
print(students)
students.update(["Ram","chitra","vani","devi"])
print(students)
```

```python
# remove an item from the set
students.remove("priya")
print (students)
```

```python
# create a set and clear list of item
students_info={"Amit",
               "Karan",
               "Riya"}

print(students_info)
students_info.clear()
print(students_info)
```

```python
# set operations
python_students={"Amit","Neha","Sanjiv","Manish"}
java_students={"Amit","Karan","DhIllip","Manish","Neha"}

# for union operation by using set
result=python_students.union(java_students)
print("Union Operation:",result)

# for intersection operation by using set
result_intersection=python_students.intersection(java_students)
print("Intersection Operation:",result_intersection)

# to find difference
result_difference=python_students.difference(java_students)
print("Difference Operation:",result_difference)

```

```python
# convert list to set
stud_id=[101,102,103,104,105]
unique_id=set(stud_id)
print(unique_id)
print(unique_id)
uniq_roll_no=list(unique_id)
print(uniq_roll_no)
```
